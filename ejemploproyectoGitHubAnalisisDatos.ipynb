***Utilizaremos Pandas y Matplotlib***

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Crear un DataFrame de ejemplo
data = {'Mes': ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo'],
        'Ventas': [2000, 3000, 4000, 3500, 5000]}
```

```python
df = pd.DataFrame(data)
```

```python
# Mostrar los datos
print(df)

# Graficar los datos
plt.plot(df['Mes'], df['Ventas'], marker='o', linestyle='-')
plt.xlabel('Mes')
plt.ylabel('Ventas')
plt.title('Ventas Mensuales')
plt.show()
```
